```python
%sql
--Create the bronze table
DECLARE bronze_catalog = 'dev';

USE CATALOG IDENTIFIER(bronze_catalog);

CREATE OR REFRESH STREAMING TABLE bronze.bronze_customers
AS 
SELECT * FROM STREAM dev.data.sales_customers;
```

```python
%sql
CREATE OR REFRESH STREAMING TABLE bronze.bronze_franchises
AS 
SELECT * FROM STREAM dev.data.sales_franchises;
```

```python
%sql
CREATE OR REFRESH STREAMING TABLE bronze.bronze_transactions
AS 
SELECT * FROM STREAM dev.data.sales_transactions;
```

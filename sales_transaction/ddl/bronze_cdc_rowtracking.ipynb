```python
%sql
DECLARE bronze_catalog = 'dev';

USE CATALOG IDENTIFIER(bronze_catalog);
```

```python
%sql

EXPLAIN CREATE MATERIALIZED VIEW IDENTIFIER(bronze_catalog || '.bronze.explain_check') 
AS 
SELECT t.* 
   ,f.*
   ,s.*
    --    t.transactiondate,
    --    f.franchiseid,
    --    s.supplierid,
    --    t.amount
FROM  IDENTIFIER(bronze_catalog || '.bronze.bronze_transactions') t
 INNER JOIN IDENTIFIER(bronze_catalog || '.bronze.bronze_customers') c   
 ON t.customerid = c.customerid
INNER JOIN IDENTIFIER(bronze_catalog || '.bronze.bronze_franchises') f 
 ON t.franchiseid = f.franchiseid
 inner join IDENTIFIER(bronze_catalog || '.bronze.bronze_suppliers') s
 ON s.supplierid = f.supplierid
```
